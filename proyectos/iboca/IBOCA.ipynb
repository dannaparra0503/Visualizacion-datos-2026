```python
# ==========================================
# PASO 2: LIMPIAR LOS DATOS
# ==========================================

# Convertimos la fecha correctamente
df["Fecha"] = pd.to_datetime(
    df.iloc[:, 0],
    errors="coerce"
)

# Reemplazamos "Sin data" por valores vacíos
df = df.replace("Sin data", np.nan)

# Lista para guardar los datos organizados
datos_limpios = []

# Recorremos todas las columnas
for columna in df.columns[1:]:

    # Obtenemos el nombre de la estación
    estacion = columna[0]

    # Obtenemos el tipo de dato
    tipo = columna[2]

    # Solo nos interesa Concentración
    if tipo == "Concentración":

        # Creamos una pequeña tabla
        temporal = pd.DataFrame({
            "Fecha": df["Fecha"],
            "Estacion": estacion,
            "PM2.5": pd.to_numeric(df[columna], errors="coerce")
        })

        # La agregamos a la lista
        datos_limpios.append(temporal)


# Unimos todas las estaciones
df_pm25 = pd.concat(datos_limpios, ignore_index=True)


# Eliminamos datos que no tengan fecha o PM2.5
df_pm25 = df_pm25.dropna(
    subset=["Fecha", "PM2.5"]
)


# ==========================================
# REVISAR EL RESULTADO
# ==========================================

print("Tamaño de la nueva tabla:")
print(df_pm25.shape)

print("\nPrimeras filas:")
display(df_pm25.head(10))

print("\nEstaciones encontradas:")
print(df_pm25["Estacion"].unique())

print("\nInformación:")
print(df_pm25.info())
```

```python
# ==========================================
# PASO 3: INDICADORES DEL DASHBOARD
# ==========================================

# 1. Promedio general de PM2.5
promedio_pm25 = df_pm25["PM2.5"].mean()

# 2. Valor máximo registrado
maximo_pm25 = df_pm25["PM2.5"].max()

# 3. Registro donde ocurrió el máximo
registro_maximo = df_pm25.loc[
    df_pm25["PM2.5"].idxmax()
]

# 4. Promedio de PM2.5 por estación
promedio_estacion = (
    df_pm25
    .groupby("Estacion")["PM2.5"]
    .mean()
    .sort_values(ascending=False)
)

# 5. Estación con mayor promedio
estacion_mayor = promedio_estacion.index[0]

# 6. Valor promedio de esa estación
valor_estacion_mayor = promedio_estacion.iloc[0]


# ==========================================
# MOSTRAR LOS RESULTADOS
# ==========================================

print("===================================")
print("       INDICADORES PM2.5")
print("===================================")

print(
    "Promedio general de PM2.5:",
    round(promedio_pm25, 2),
    "µg/m³"
)

print(
    "Máximo registrado:",
    maximo_pm25,
    "µg/m³"
)

print(
    "Estación con mayor promedio:",
    estacion_mayor
)

print(
    "Promedio de esa estación:",
    round(valor_estacion_mayor, 2),
    "µg/m³"
)

print(
    "Fecha y hora del máximo:",
    registro_maximo["Fecha"]
)
```

```python
# ==========================================
# PASO 4: PROMEDIO DE PM2.5 POR ESTACIÓN
# ==========================================

# Calculamos el promedio de PM2.5 para cada estación
promedio_estacion = (
    df_pm25
    .groupby("Estacion")["PM2.5"]
    .mean()
    .sort_values(ascending=False)
)

# Mostramos los resultados
print("PROMEDIO DE PM2.5 POR ESTACIÓN:")
print(promedio_estacion.round(2))
```

```python
# ==========================================
# GRÁFICO: PM2.5 POR ESTACIÓN
# ==========================================

fig = px.bar(
    x=promedio_estacion.index,
    y=promedio_estacion.values,
    labels={
        "x": "Estación",
        "y": "PM2.5 (µg/m³)"
    },
    title="Promedio de PM2.5 por estación"
)

fig.update_layout(
    xaxis_tickangle=-45
)

fig.show()
```

```python
# ==========================================
# PASO 5: PM2.5 A LO LARGO DEL TIEMPO
# ==========================================

# Calculamos el promedio de PM2.5 para cada fecha
promedio_fecha = (
    df_pm25
    .groupby("Fecha")["PM2.5"]
    .mean()
    .reset_index()
)

# ==========================================
# CREAR EL GRÁFICO
# ==========================================

fig = px.line(
    promedio_fecha,
    x="Fecha",
    y="PM2.5",
    labels={
        "Fecha": "Fecha y hora",
        "PM2.5": "PM2.5 (µg/m³)"
    },
    title="Evolución del PM2.5 a lo largo del tiempo"
)

fig.show()
```

```python
# ==========================================
# PASO 6: PM2.5 SEGÚN LA HORA
# ==========================================

# Creamos la columna Hora a partir de Fecha
df_pm25["Hora"] = df_pm25["Fecha"].dt.hour


# Calculamos el promedio de PM2.5 para cada hora
promedio_hora = (
    df_pm25
    .groupby("Hora")["PM2.5"]
    .mean()
    .reset_index()
)

# Ordenamos las horas
promedio_hora = promedio_hora.sort_values("Hora")


# ==========================================
# MOSTRAR LOS RESULTADOS
# ==========================================

print("PROMEDIO DE PM2.5 SEGÚN LA HORA:")

display(promedio_hora.round(2))


# ==========================================
# CREAR EL GRÁFICO
# ==========================================

fig = px.line(
    promedio_hora,
    x="Hora",
    y="PM2.5",
    markers=True,
    labels={
        "Hora": "Hora del día",
        "PM2.5": "PM2.5 (µg/m³)"
    },
    title="Promedio de PM2.5 según la hora del día"
)

fig.show()
```

```python
# ==========================================
# PASO 7: PM2.5 POR DÍA DE LA SEMANA
# ==========================================

# Creamos la columna con el día de la semana
df_pm25["Dia_semana"] = df_pm25["Fecha"].dt.day_name()


# Calculamos el promedio de PM2.5 para cada día
promedio_dia = (
    df_pm25
    .groupby("Dia_semana")["PM2.5"]
    .mean()
    .reset_index()
)


# ==========================================
# ORDENAR LOS DÍAS CORRECTAMENTE
# ==========================================

orden_dias = [
    "Monday",
    "Tuesday",
    "Wednesday",
    "Thursday",
    "Friday",
    "Saturday",
    "Sunday"
]

promedio_dia["Dia_semana"] = pd.Categorical(
    promedio_dia["Dia_semana"],
    categories=orden_dias,
    ordered=True
)

promedio_dia = promedio_dia.sort_values("Dia_semana")


# ==========================================
# MOSTRAR LOS RESULTADOS
# ==========================================

print("PROMEDIO DE PM2.5 POR DÍA DE LA SEMANA:")

display(promedio_dia.round(2))


# ==========================================
# CREAR EL GRÁFICO
# ==========================================

fig = px.bar(
    promedio_dia,
    x="Dia_semana",
    y="PM2.5",
    labels={
        "Dia_semana": "Día de la semana",
        "PM2.5": "PM2.5 (µg/m³)"
    },
    title="Promedio de PM2.5 por día de la semana"
)

fig.show()
```

```python
# ==========================================
# PASO 8: MAYOR Y MENOR PM2.5 POR ESTACIÓN
# ==========================================

# Calculamos el promedio de PM2.5 por estación
promedio_estacion = (
    df_pm25
    .groupby("Estacion")["PM2.5"]
    .mean()
    .sort_values(ascending=False)
)


# ==========================================
# ESTACIÓN CON MAYOR PROMEDIO
# ==========================================

estacion_mayor = promedio_estacion.idxmax()
valor_mayor = promedio_estacion.max()


# ==========================================
# ESTACIÓN CON MENOR PROMEDIO
# ==========================================

estacion_menor = promedio_estacion.idxmin()
valor_menor = promedio_estacion.min()


# ==========================================
# MOSTRAR RESULTADOS
# ==========================================

print("===================================")
print("   ANÁLISIS POR ESTACIÓN")
print("===================================")

print(
    "Estación con mayor promedio:",
    estacion_mayor
)

print(
    "Promedio más alto:",
    round(valor_mayor, 2),
    "µg/m³"
)

print()

print(
    "Estación con menor promedio:",
    estacion_menor
)

print(
    "Promedio más bajo:",
    round(valor_menor, 2),
    "µg/m³"
)
```

```python
# ==========================================
# PASO 9: SELECCIONAR UNA ESTACIÓN
# ==========================================

# Lista de estaciones disponibles
estaciones = sorted(df_pm25["Estacion"].unique())

print("ESTACIONES DISPONIBLES:")

for i, estacion in enumerate(estaciones, start=1):
    print(i, "-", estacion)


# ==========================================
# ELEGIR UNA ESTACIÓN
# ==========================================

numero = int(input("\nEscribe el número de la estación que quieres consultar: "))

estacion_seleccionada = estaciones[numero - 1]

print("\nEstación seleccionada:", estacion_seleccionada)


# ==========================================
# FILTRAR LOS DATOS
# ==========================================

datos_estacion = df_pm25[
    df_pm25["Estacion"] == estacion_seleccionada
].copy()


print("\nCantidad de registros:")
print(len(datos_estacion))

print("\nPrimeros registros:")
display(datos_estacion.head(10))
```

```python
# ==========================================
# PASO 10: GRÁFICA DE LA ESTACIÓN
# ==========================================

fig = px.line(
    datos_estacion,
    x="Fecha",
    y="PM2.5",
    labels={
        "Fecha": "Fecha y hora",
        "PM2.5": "PM2.5 (µg/m³)"
    },
    title=f"Evolución del PM2.5 - {estacion_seleccionada}"
)

fig.show()
```

```python
# ==========================================
# PASO 11: RANKING DE ESTACIONES
# ==========================================

# Ordenamos las estaciones de mayor a menor
ranking_estaciones = (
    df_pm25
    .groupby("Estacion")["PM2.5"]
    .mean()
    .sort_values(ascending=False)
    .reset_index()
)

# Agregamos la posición del ranking
ranking_estaciones["Posición"] = range(
    1,
    len(ranking_estaciones) + 1
)

# Organizamos las columnas
ranking_estaciones = ranking_estaciones[
    ["Posición", "Estacion", "PM2.5"]
]


# ==========================================
# MOSTRAR EL RANKING
# ==========================================

print("RANKING DE ESTACIONES SEGÚN PROMEDIO DE PM2.5:")

display(
    ranking_estaciones.round(2)
)
```
